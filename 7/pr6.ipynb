## Q5. Given vector a = [1,2,3,4,5,6], compute its Euclidean norm manually and using NumPy.

Hint: By "manually," I mean solving it step by step using only basic NumPy functions.

````python
a = np.array([1, 2, 3, 4, 5, 6])

norm1 = np.sqrt(np.sum(a**2))
norm2 = np.linalg.norm(a)
print(norm1, norm2, sep='\n')
````

## Q7. Compute the dot product of vectors

a = [1,2,3,4,5,6] & b = [0,-4,3,-3,6,5]
using:
1. Manual loop
2. NumPy np.dot() 

````python
a = np.array([1,2,3,4,5,6])
b = np.array([0,-4,3,-3,6,5])
d1 = 0
d2 = 0
for i in range(len(a)):
    d1 += a[i] * b[i]
d2 = np.dot(a, b)
print(d1, d2)
````

## Q9: Compute the Euclidean distance matrix between every point in group A and every point in group B.


````python
A = np.array([
[ 2, 1],
[ 5, 4],
[ 1, 6],
[ 7, 2]
])
B = np.array([
[0, 0],
[6, 3],
[3, 5]
])
print(A.shape, B.shape, sep='\n')

A1 = A[:, np.newaxis, :]
B1 = B[np.newaxis,:, :]
dif = A1 - B1
print(dif)
print(A1.shape, B1.shape, dif.shape, sep='\n')

sum_sq = np.sum(dif**2, axis=2)
dist_matrix = np.sqrt(sum_sq)

print(dist_matrix.shape)
print(dist_matrix)
````
