Kerala Flood Data

Data of monthly rainfall in Kerala from 1901 to 2018 is provided. It also contains information of floods.

Columns in Dataset:
- ```YEAR```: Year between 1901 to 2018
- ```JAN, FEB,..., DEC```:Rainfall in mm in the specified month
- ```ANNUAL RAINFALL```: Total rainfall in mm in that year
- ```FLOODS```: YES or NO, in that yea

````python
df2 = pd.read_csv('kerala_floods.csv')
print(df2.shape)
df2.head()
````

````python
df2.isna().sum()
````

We need to use Bayes Therorm to calculate the probability of having a rainfall more than 500 mm in month
of June, provided it has flooded in that year ?

## 1- We may start by preparing the required columns:

- Create a new column named "June_more_than_500" and assign a value of 1 if the amount is more than 500 and 0 if it is less
than 500.
- Do the same for the 'FLOODS' column, assigning 1 for 'yes' and 0 for 'no'

````python
df2['June_more_than_500'] = (df2['JUN'] > 500).astype(int)
df2['June_more_than_500'].head()
````

````python
df2['FLOODS2'] = (df2['FLOODS'] == 'YES').astype(int)
df2['FLOODS2'].head()
````

## 2- Select only the columns that will help to answer the main question.

````python
df2[['June_more_than_500', 'FLOODS2']].head()
````

## 3- Create a cross tabulation for 'FLOODS' and 'June_more_than_500'

````python
pd.crosstab(df2['FLOODS2'], df2['June_more_than_500'], margins=True)
````

## 4- What is the probability of a rainfall more than 500mm in June?

````python
p_rain = df2['June_more_than_500'].mean()
print(p_rain)
````

## 5- What is the probability of having a flood?

````python
df2['FLOODS2'].mean()
````

## 6- What is the probability of having a flood and a rainfall more than 500mm in June?

````python
p_f_and_r = ((df2['FLOODS2'] == 1) & (df2['June_more_than_500'] == 1)).mean()
print(p_f_and_r)
````

## 7- What is the Probability of having a flood given it rained more than 500mm in June?

````python
p_flood_given_rain = p_f_and_r / p_rain
print(p_flood_given_rain)
````

## 8- What is the probability of a rainfall less than 500mm in June?

````python
p_not_rain = 1 - p_rain
print(p_not_rain)
````

## 9- What is probability of having a flood given that it rained less than 500mm?

````python
p_flood_given_not_rain = ((df2['FLOODS2'] == 1) & (df2['June_more_than_500'] == 0)).mean() / p_not_rain
print(p_flood_given_not_rain)
````

## Question: Calculate the probability of having a rainfall more than 500mm in month of June, provided it has flooded in that year ?

### 10- Answer this question by using Bayes Theorem and the law of the Total Probaility

p(B)= P(b|a)*p(a)+p(b|~a)*p(~a) flood total

````python
p_flood_total = (p_flood_given_rain * p_rain) + (p_flood_given_not_rain * p_not_rain)
(p_flood_given_rain * p_rain) / p_flood_total
````

### 11- Answer this question by using Bayes Theorem only.

p(a|b) = (p(b|a) * p(a)) / p(b)

````python
p_a = (df2['JUN'] > 500).mean() 
p_b = (df2['FLOODS'] == 'YES').mean() 
p_b_given_a = (df2[df2['JUN'] > 500]['FLOODS'] == 'YES').mean()
(p_b_given_a * p_a) / p_b
````
