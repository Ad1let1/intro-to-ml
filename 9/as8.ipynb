### Q1: How many null values are there in each column?

````python
df.isna().sum()
````

````python
df['wheelbase'] = df['wheelbase'].fillna(df['wheelbase'].max())
````

````python
df['carwidth'] = df['carwidth'].fillna(df['carwidth'].min())
````

````python
df['carheight'] = df['carheight'].interpolate(method='linear')
````

````python
df['enginesize'] = df['enginesize'].fillna(df['enginesize'].mode()[0])
df.isna().sum()
````

````python
Pb = (df.fueltype == 'gas').mean()
print(Pa * Pb )
````

````python
(Pa * Pb) / Pb
````

### Q5: Given that the car height is higer than 50, find the probability that the car engine size is greater than 130?

````python
# P(sz > 130 | height > 50)
Pa = (df.enginesize > 130).mean()
Pb = (df.carheight > 50).mean()
(Pa * Pb) / Pb
````

### Q7: What is the probability that the salary is greater than 20000 and less than 30000 ?

To answer this question, you should do the following:
1. Identify the distribution by
    - Plotting Histogram
    - calculate the skewness
    - Which distribution does this data follow?
2. Compute mean and std

3. Convert to Z-scores
4. Use CDF difference with the use of the right distribution function from scipy.stats

#### 1. Distribution

````python
sns.histplot(df2, x='salary', kde=True)
````

````python
salary_skew = df2['salary'].dropna().skew()
print(salary_skew)
````

Since the -0.5 < salary_skew < 0.5, it is normally distributed

#### 2. mean, std

````python
print(df2['salary'].mean())
print(df2['salary'].std())
````

#### 3. Converting to Z score

````python
mu = df2['salary'].mean()
sigma = df2['salary'].std()

df2['salary_z'] = (df2['salary'] - mu) / sigma
df2[['salary', 'salary_z']].head()
````

#### 4. CDF difference with the use of the right distribution function from scipy.stats

````python
from scipy.stats import norm

lower_bound = 20000
upper_bound = 30000

z1 = (lower_bound - mu) / sigma
z2 = (upper_bound - mu) / sigma
p = norm.cdf(z2) - norm.cdf(z1)
p
````

P(20000< salary < 30000) =  9.8%
