## Question 3:

```python
v = np.array([4, 5])
```

```python
plt.plot([0, v[0]], [0, v[1]], label='x')
plt.plot([0,0], [5,-3], 'k--')
plt.plot([5,-3], [0,0], 'k--')
plt.grid()
plt.axis((-1,6,-1,6))
plt.legend()
plt.show()
```

## Question 4:

```python
v3 = np.array([5,-4,3])
plt.figure(figsize = plt.figaspect(1))
plt.axes(projection = '3d')
plt.plot([0,v3[0]], [0, v3[1]], [0, v3[2]])
plt.plot([0, 0],  [-5, 5], [0, 0], 'k--',label = 'x')
plt.plot([0, 0],  [0, 0],  [-5, 5],'k--', label = 'y')
plt.plot([-1,4], [0, 0],  [0, 0], 'k--', label = 'z')
plt.legend()
plt.show()
```

## Question 5:

```python
v1 = np.array([4, -2])
v2 = np.array([3, 5])
v3 = v1+v2
print(v3)
plt.plot([0, v1[0]],[0, v1[1]],'b',label='v1')
plt.plot([0, v2[0]]+v1[0],[0, v2[1]]+v1[1],'r',label='v2')
plt.plot([0, v3[0]],[0, v3[1]],'g',label='v1+v2')
plt.plot([-7, 7],[0, 0],'k--') 
plt.plot([0, 0],[-7, 7],'k--') 
plt.legend()
plt.axis('square')
plt.axis((-8, 8, -8, 8 ))
plt.grid()
plt.show()
```
